# Giải thích về Managed Tables trong Databricks

## 1️⃣ Managed Tables là gì?
Managed Table là bảng do Databricks quản lý hoàn toàn. Khi bạn tạo một bảng managed, dữ liệu được sao chép vào kho lưu trữ của Databricks.

Điều này khác với External Tables, nơi dữ liệu không được sao chép mà chỉ được tham chiếu từ nguồn bên ngoài.

## 2️⃣ Cách hoạt động của Managed Tables
Khi bạn tạo bảng managed, Databricks sẽ lưu trữ dữ liệu trong metastore nội bộ.

Nếu bạn xóa bảng managed, dữ liệu gốc cũng sẽ bị xóa.

# Clone our sales_customers_V Table

Let's create a deep clone of our sales_customers_V table. A deep clone will create an independent copy of our table that we can use to clean the data.

```python
%sql
CREATE TABLE IF NOT EXISTS sales_customers_V
  DEEP CLONE sales_customers_V;
SELECT * FROM samples.bakehouse.sales_customers;
```
